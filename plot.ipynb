solving equation x+3z+2y+10w=10 with y(0,5),z(20,50),w(0,1),u(5,10),v(20,25)

```python
import matplotlib.pyplot as plt
import numpy as np

# Define your requested ranges for x, y, and w
x = np.linspace(2, 10, 50)
y = np.linspace(0, 5, 50)
w_values = [0.0, 0.5, 1.0]  # Slices across w in [0, 1] , each plane (3 plane for 3 values of w)

X, Y = np.meshgrid(x, y)

fig = plt.figure(figsize=(9, 6))
ax = fig.add_subplot(111, projection="3d")

for w_val in w_values:
  # Solve for z: x + 2y + 3z + 10w = 10
  Z = (10 - 10 * w_val - X - 2 * Y) / 3.0
  ax.plot_surface(X, Y, Z, alpha=0.6, label=f"w = {w_val}")

ax.set_xlim(2, 10)
ax.set_ylim(0, 5)
# Note: Within x in [2,10], y in [0,5], w in [0,1], actual z values lie in [-10, 2.67].
# Uncomment the line below if you strictly want the z-axis view fixed to [20, 50] (the plot will appear empty):
# ax.set_zlim(20, 50)

ax.set_title("Plane Slices for x ∈ [2, 10], y ∈ [0, 5], w ∈ [0, 1]")
ax.set_xlabel("x (2 to 10)")
ax.set_ylabel("y (0 to 5)")
ax.set_zlabel("z")
plt.tight_layout()
plt.show()
```
